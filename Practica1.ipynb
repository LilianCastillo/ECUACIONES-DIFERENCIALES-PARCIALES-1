```python
import matplotlib.pyplot as plt
import numpy as np

# Generate x values using numpy.linspace for smoother curve
x = np.linspace(-10, 10, 100)

# Calculate y values using the equation y = x^2
y = x**2

# Create the plot
plt.plot(x, y)

# Customize the plot
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Graph of $y = x^2$")

# Display the plot
plt.show()
```
